```python
qubits = [cirq.q(5,1), cirq.q(5,2)]

circuit = cirq.Circuit(cirq.H.on_each(qubits), cirq.CZ(*qubits), cirq.H(qubits[1]), cirq.M(*qubits, key='m'))
```

```python
circuit
```

```python
sampler = processer.get_sampler() #specify device config name or revision, circuits per job (defult 1)
```

```python
result = sampler.run(circuit, repetitions=100)
```
